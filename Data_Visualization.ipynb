<span style='font-size: 17px'><b>Graph 3: Cumulative Sum (Random Walk)

* <b>Objective:</b> Plot the cumulative sum of the random data points using 'y.cumsum()'.
* <b>Key Takeaway:</b> Transforms independent random data into a 'Random Walk.' This illustrates how small, random daily changes accumulate into a long-term price trend.</span>

```python
plt.plot(y.cumsum())
plt.grid(False)
plt.axis('equal')
plt.show()
```

<span style='font-size:17px'><b>Graph 4: Equal Axis Scaling</b>

* <b>Objective:</b> Use 'plt.axis('equal')' and 'plt.grid(False)' to modify the plot's proportions and background.
* <b>Key Takeaway:</b> Forces the x and y axes to use the same scale. This reveals the 'true' flat nature of the cumulative data, preventing the trend from appearing artifically steep or exaggerated.</span>

```python
plt.plot(y.cumsum())
plt.xlim(-1,20)
plt.ylim(np.min(y.cumsum())-1,
         np.max(y.cumsum())+1);
plt.show()
```

<span style='font-size:17px'><b>Graph 5: Manual Axis Scaling</b>

<span style='font-size:17px'><b>Graph 17: Boxplots for Statistical Summaries</b>

* <b>Objective:</b> Use 'plt.boxplot()' to visualize the quartiles and outliers of the two data series.
* <b>Key Takeaway:</b> Provides a concise view of the data's dispersion. The central 'box' represents the middle 50% of the data, while the 'whiskers' and circular markers highlight the extreme fluctuations. This makes it easy to compare the typical range and the frequency of high-impact volatility events between the two datasets. </span>

```python
def func(x):
    return 0.5 * np.exp(x)+1 
a, b = 0.5, 1.5
x = np.linspace(0,2)
y = func(x)
Ix = np.linspace(a,b)
Iy = func(Ix)
verts = [(a, 0)] + list(zip(Ix, Iy)) + [(b,0)]
```

```python
from matplotlib.patches import Polygon
fig, ax = plt.subplots(figsize=(10,6))
plt.plot(x, y, 'b', linewidth=2)
plt.ylim(bottom=0)
poly = Polygon(verts, facecolor='0.7',edgecolor='0.5')
ax.add_patch(poly)
plt.text(0.5* (a+b), 1, r'$\int_a^b f(x)\mathrm{d}x$', 
         horizontalalignment='center',fontsize=20)
plt.figtext(0.9, 0.075, '$x$')
plt.figtext(0.075, 0.9, '$f(x)$')
ax.set_xticks((a, b))
ax.set_xticklabels(('$a$', '$b$'))
ax.set_yticks([func(a), func(b)])
ax.set_yticklabels(('$f(a)$', '$f(b)$'))
plt.show()
```

<span style='font-size:17px'><b>Graph 18: Mathematical Shading and LaTeX Annotations</b>

* <b>Objective:</b> Use 'matplotlib.patches.Polygon' to shade the area under a function and annotate it with LaTeX mathematical symbols.
* <b>Key Takeaway:</b> Demonstrates the visualization of calculus concepts. The shaded region represents the integral of the function, providing a geometric view of the total 'accumulated' value. This allows us to see not just the volatility of the curve, but the specific magnitude of the area between two defined points.  </span>

```python
strike = np.linspace(50, 150, 24)
```

```python
ttm = np.linspace(0.5, 2.5, 24)
```

```python
strike, ttm = np.meshgrid(strike, ttm)
```

```python
strike[:2].round(1)
```

```python
iv = (strike - 100) ** 2 / (100 * strike) / ttm
```

```python
iv[:5, :3]
```

```python
from mpl_toolkits.mplot3d import Axes3D
fig = plt.figure(figsize=(10,6))
ax = fig.add_subplot(111, projection='3d')
surf = ax.plot_surface(strike, ttm, iv, rstride=2, cstride=2, 
                       cmap=plt.cm.coolwarm, linewidth=0.5, 
                       antialiased=True)
ax.set_xlabel('strike')
ax.set_ylabel('time-to-maturity')
ax.set_zlabel('implied volatility')
fig.colorbar(surf, shrink=0.5, aspect=5)
plt.show()
                       
```

<span style='font-size:17px'><b>Graph 23: Technical Analysis with Candletstick Charts</b>

* <b>Objective:</b> Utilize 'cf.QuantFig' to generate a profressional-grade candlestick chart for exchange rate data.
* <b>Key Takeaway:</b> Provides a high-density view of market dynamics. Unlike a simple line, each candle captures the full range of price action, making it easier to identify reversal patters or periods of extreme intraday instability that a standard closing-price plot would miss. </span>

```python
qf.add_bollinger_bands(periods=15,
                       boll_std=2)
```

```python
import plotly.graph_objects as go

fig = go.Figure(data=[go.Candlestick(x=quotes.index,
                open=quotes['AskOpen'],
                high=quotes['AskHigh'],
                low=quotes['AskLow'],
                close=quotes['AskClose'],
                name='EUR/USD')])

sma = quotes['AskClose'].rolling(window=15).mean()
std = quotes['AskClose'].rolling(window=15).std()

upper_band = sma + (std * 2)
lower_band = sma - (std * 2)

fig.add_trace(go.Scatter(x=quotes.index, y=upper_band, name='Upper Band', line=dict(color='rgba(173, 216, 230, 0.5)')))
fig.add_trace(go.Scatter(x=quotes.index, y=lower_band, name='Lower Band', line=dict(color='rgba(173, 216, 230, 0.5)'), fill='tonexty'))
fig.add_trace(go.Scatter(x=quotes.index, y=sma, name='SMA 15', line=dict(color='gray', dash='dash')))

fig.update_layout(
    title='EUR/USD Analysis: Candlesticks & Bollinger Bands',
    template='plotly_dark',
    width=1000,
    height=600,
    xaxis_rangeslider_visible=False 
)

fig.show()
```
