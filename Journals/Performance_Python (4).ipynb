<span style='font-size:17px'><b>The Iterative Loop: Speeding into a Wall</b>

* <b>Logic Efficiency:</b> By switching to a Iterative Algorithm ('for i in range'), the 'forgetfulness' of the recursion is fixed. The computer nows only does $N$ steps of math instead of millions, which is why the time dropped to microseconds.
* <b>The 'Box' Problem:</b> While the loop is fast, the variables ($x$ and $y$) are like container. In Cython and Numba, these containers are high-speed '64-bit boxes' that can only hold a certain amount of data.
* <b>The 63-Bit Overflow:</b> Once your iterative loop reaches $Fib(93)$, the number bceomes too large for a 64-bit box. The extra data 'spills over,' causing the bits to flip and creating the negative numbers or 63-bit cutoff seen in the results.
* <b>Why Python Wins:</b> Standard Python's iterative loop is slightly slower because it uses 'Expanding Tanks' instead of 'Fixed Boxes.' It waits for the number to grow and automatically gives it more memory, which is how it reaches 103 bits for $Fib(150)$.

```python
%%cython
from libc.math cimport sqrt  
import random

def mcs_pi_cy1(int n):
    cdef int i, circle = 0
    cdef double x, y
    for i in range(n):
        x = random.random()
        y = random.random()
        if sqrt(x*x + y*y) <= 1.0:
            circle += 1
    return (4.0 * circle) / n
```

```python
%%cython -a
from libc.stdlib cimport rand, RAND_MAX
from libc.math cimport sqrt

def mcs_pi_cy2(int n):
    cdef int i, circle = 0
    cdef double x, y
    for i in range(n):
        # The <double> here is the magic fix
        x = rand() / <double>RAND_MAX
        y = rand() / <double>RAND_MAX
        if sqrt(x*x + y*y) <= 1.0:
            circle += 1
    return (4.0 * circle) / n
```

<span style='font-size:17px'><b>The Monte Carlo Pi Simulation Breakdown</b>

* <b>The Geometric Core:</b> The algorithm estimates the $\pi$ by randomly 'throwing darts' at a unit square (area of 1) and checking how many land inside a quarter-circle arc (are of $\pi/4$). By calculating the ratio of 'hits' in the circle to the total number of darts thrown and multiplying by 4, we arrive at an estimate of $\pi$.
* <b>The Python Performance Gap:</b> A standard Python loop ('mcs_pi_py') is slow because the interpreter must manage high-level objects for every single iteration, which can lead to execution times of several seconds for 10 million loops. Using NumPy or Numba speeds this up significantly by performing operations in bulk or compiling them into machine code, dropping the runtime to under 600 milliseconds.
* <b>The Cython 'C-speed' Secret:</b> In your Cython version ('mcs_pi_cy2'), the code is transformed into pure C, which eliminates the heavy Python overhead. By using 'cimport' to access the C library's 'rand()' and 'sqrt()' functions and explicitly declaring variables as 'int' or 'double', the loop runs at hardware-level speeds (roughly 359 milliseconds).


```python
import math
```

```python
S0 = 36.
T = 1.0
r = 0.06
sigma = 0.2
```

```python
def simulate_tree(M):
    dt = T/M
    u = math.exp(sigma * math.sqrt((dt)))
    d = 1 / u
    S = np.zeros((M +1, M + 1))
    S[0,0] = S0
    z = 1
    for t in range(1, M + 1):
        for i in range(z):
            S[i, t] = S[i, t-1] * u
            S[i+1, t] = S[i, t-1] * d 
        z += 1 
    return S
```

<span style='font-size:17px'><b>The High-Performance Binomial Tree Breakdown</b>

* <b>The Vectorization Shift:</b> While the original 'simulate_tree' used nested loops, the NumPy version ('simulate_tree_np') uses broadcasting and matrix math to calculate every possible price at once. By replacing the loops with 'up - down' index logic, the execution for 500 steps dropped from <b>167 ms</b> to a staggering <b>11.5ms</b>.
* <b>The Numba 'Magic':</b> By simply wrapping the original function in the '@numba.jit' decorator, the computer was able to compile the Python loops into machine code. This provides a massive speed boost (reducing wall time to under 1 ms) without having to rewrite a single line of the original logic.
* <b>Cython's Static Typing:</b> In the 'simulate_tree_cy' version, we explicitly declared variables like 'cdef double S0' and 'cdef int z, t, i'. This tells the computer exactly how much memory to set aside upfront, allowing it to bypass the slow 'type-checking' that makes standard Python sluggish.
* <b>C-Standard Math Integration:</b> By importing 'exp' and 'sqrt' directly from 'libc.math', the simulation avoids calling high-level Python math objects. This ensures that the most computationally expensive operations are performed using optimized C libraries, resulting in a line 4.61 ms runtime.
* <b>Efficient Recombining Structure:</b> The algorithm utilizes a 'recombining' tree where an upward move followed by a downward move returns to the same point. This prevents the number from exploding exponentially, ensuring the model remains computationally scalable even as the number of time steps increases.

```python
def mcs_simulation_py(p):
    M, I = p
    dt = T / M
    S = np.zeros((M + 1, I))
    S[0] = S0
    rn = np.random.standard_normal(S.shape)
    for t in range(1, M + 1):
        for i in range(I):
            S[t, i] = S[t-1, i]* math.exp((r - sigma ** 2/ 2) * dt+
                                          sigma * math.sqrt(dt) * rn[t, i])
    return S
```

<span style='font-size:17px'><b>The Euler Scheme: Discretization & Implementation</b>

* <b>The Difference Equation:</b> The code implements the Euler Scheme to approximate the Continuous-Time BSM equation by breaking it into discrete time steps ($\Delta t$). This allows the computer to iteratively calculate the next stock price by adding a fixed drift and a random stochastic shock to the current price.
* <b>Log-Normal Distribution Correction:</b> By using the exponential form $e^{(r - 0.5 \ sigma^2)}$,the simulation ensures that the stock price remains positive and follows a log-normal distribution. The $-0.5\ sigma^2$ term is a critical adjustment by Ito's Lemma to ensure the expected value of the simulation matches the theoretical risk-neutral growth.
* <b>Payoff Valuation:</b> The final calculation of $C0$ (3.85) demonstrates the power of the Monte Carlo method under the Euler Scheme. By taking the average of all terminal payoffs and discounting them back to the present at the risk-free rate, the model provides an accurate numerical estimate of the option's fair market value.

```python
%%cython
import numpy as np
cimport numpy as np
cimport cython
from libc.math cimport exp, sqrt
cdef float S0 = 36.
cdef float T = 1.0
cdef float r = 0.06
cdef float sigma = 0.2
@cython.wraparound(False)
def mcs_simulation_cy(p):
    cdef int M,I
    M, I = p 
    cdef float dt = T / M
    cdef double [:, :] S = np.zeros((M + 1, I))
    cdef double [:, :] rn = np.random.standard_normal((M + 1, I))
    cdef int t, i
    S[0]
    for t in range(1, M + 1):
        for i in range(I):
            S[t, i] = S[t-1, i] * exp((r - sigma ** 2/2) * dt +
                                      sigma * sqrt(dt) * rn[t, i])
    return np.array(S)         
```
