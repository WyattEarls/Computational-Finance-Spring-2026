```python
import random
```

```python
def average_py(n):
    s = 0
    for i in range(n):
        s += random.random()
    return s / n 
```

```python
n = 10000000
```

```python
%time average_py(n)
```

```python
%timeit average_py(n)
```

```python
%time sum([random.random() for _ in range(n)]) / n 
```

<span style='font-size:17px'><b>Basic Python Benchmarking</b>

* <b>Function Definition:</b> Creates a pure Python function 'average_py' that uses a simple 'for' loop to calculate the mean of $n$ random numbers.
* <b>Basline Performance:</b> Benchmarks the function with 10 million iterations, recording a wall time of 1.31 seconds for a single execution.
* <b>Statistical Timing:</b> Uses '%timeit' to provide a more robust average, showing approximately 1.36 seconds per loop across 7 different runs.
* <b>List Comprehension Comparison:</b> Tests a one-liner approach using 'sum()' on a list comprehension, which performs slower at 2.17 seconds due to memory overhead.

<span style='font-size:17px'><b>NumPy Vectorization</b>

* <b>Function Optimization:</b> Implements 'average_np' using NumPy's 'random' and 'mean' methods to replace the manual Python loop.
* <b>Syntax Debugging:</b> Identifies that 's.mean' returns the function object whereas 's.mean()' is required to execute the calculation.
* <b>Performance Leap:</b> Acheives a wall time of 133 ms for 10 million numbers, making it roughly 10x faster than the pure Python loop.
* <b>Vectorized Execution:</b> Demonstrates how NumPy processes the entire arra at once in C, drastically reducing the overhead compared to Python iteration.

<span style='font-size:17px'><b>Numba Just-In-Time (JIT) Compilation</b>

* <b>Compilation Overhead:</b> Incurs a one-time 'warm-up' cost during the first execution as Numba translates Python bytecode into optimized machine code.
* <b>Execution Efficiency:</b> Reduces execution time from ~1.3 seconds to 119 milliseconds once the function is cached in its compiled state.
* <b>Seamless Integration:</b> Demonstrates the 'numba.jit()' wrapper's ability to take a standard Python function and optimize it without manual code rewrites.
* <b>Performance Parity:</b> Acheives results comparable to NumPy's vectorized methods by running calculations directly at the CPU instruction level.

```python
%%cython 
from libc.stdlib cimport rand 
cdef extern from 'limits.h':
    int INT_MAX
cdef int i
cdef float rn
for i in range(5):
    rn = rand() / INT_MAX
    print(rn)
```

```python
%%cython -a 
from libc.stdlib cimport rand
cdef extern from 'limits.h':
    int INT_MAX
def average_cy2(int n):
    cdef int i 
    cdef float s = 0
    for i in range(n):
        s += rand() / INT_MAX
    return s / n
```

```python
%time average_cy2(n)
```

```python
%timeit average_cy2(n)
```

<span style='font-size:17px'><b>Cython and C-Library Integration</b>

* <b>Library Transition:</b> Shifts from the high-level Python 'random' module to the low-level 'libc.stdlib' C library to eliminate interpreter overhead.
* <b>Static Declaration:</b> Uses 'cdef' to explicitly define variable types and external C constants like 'INT_MAX', allowing for direct machine-code mapping.
* <b>Optimization Analysis:</b> Employs the '-a' annotation flag to identify and remove 'yellow' Python-interaction lines, achieving a 'white' pure-C execution loop.
* <b>Efficiency Breakthrough:</b> Reduces execution time from ~2.0 seconds to 153 milliseconds by bypassing the Python virtual machine entirely during the calculation.

```python
def is_prime(I):
    if I % 2 == 0: return False
    for i in range(3, int(I ** 0.5) + 1, 2):
        if I % i == 0: return False
    return True
```

<span style='font-size:17px'><b>Computational Primality Testing & Efficiency</b>

* <b>Optimized Trial Division:</b> The 'is_prime' function implements a loop that only checks odd numbers up to the square root of the target ($\sqrt{1}$), which is a significantly more efficient approach than checking every number up to $I$.
* <b>Time Complexity Practice:</b> The results show a massive jump in 'Wall Time' as the numbers grow; while $10^8 + 3$ is processed in **$68.4 \ \ mu s$**, the 57-bit prime $p2$ requires **$15$ seconds** to verify.
* <b>Bit-Length Impact:</b> The use of 'p2.bit_length()' highlights that asthe binary size of a number increases (to **57 bits** in this case), the number of potential divisors grows, causing a linear algorithm to slow down noticeably.
* <b>Performance Monitoring:</b> By using the '%time' magic command, the output distinguishes **CPU time** (active processing) and **Wall time** (total time elapsed), which is identifying bottlenecks in cryptographic or mathematical scripts.

<span style='font-size:17px'><b>Performance Scaling and Limitations in Cython and Numba</b>

* <b>Static Typing Efficiency:</b> The transition from 'is_prime_cy1' to 'is_prime_cy2' demonstrates that explicitly declaring types (e.g., 'long I' and 'cdef long i') allows Cython to bypass Python's overhead, reducing execution time for $p1$ from **$334 \ \mu s$** results likely utilizes a 64-bit environment.
* <b>The 32-bit vs. 64-bit Bottleneck:</b> The 'OverflowError' encounterd with $p2$ occurs because the code attempts to fit a **57-bit** integer into a C 'long'; on Windows systems, this type is typically limited to 32 bits, whereas the book's successful **1.68s** result likely utilizes a 64-bit environment.
* <b>Numba's JIT advantage:</b> Numba provides a high-performance alternative to Cython with less manual configuration, successfully processing the large prime $p2$ in **$657 \ ms$** compared to the original Python time of **$15 \ seconds$**.
* <b>Compilation Latency:</b> Initial calls to Numba-decorated functions (like the first run of 'is_prime_nb') includes a one-time 'warm-up' cost for machine code compilation (**$192 \ ms$**), while subsequent calls run at near-native speeds (**$11.2 \ \mu s$**).
